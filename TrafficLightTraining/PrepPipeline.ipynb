```python
import cv2
import numpy as np
import os
```

```python
def procRPLTRlabelData(dataList, h, w):
    data = np.array(dataList)
    data = np.unique(data, axis=0)
    data = data[:, :5]
    
    print(data)
    
    labels = []
    
    for row in data:
        cls, xMin, yMin, xMax, yMax = row
    
        xCenter = ((xMin + xMax) / 2) / w
        yCenter = ((yMin + yMax) / 2) / h
        bw = (xMax - xMin) / w
        bh = (yMax - yMin) / h
    
        labels.append(f"{int(cls)} {xCenter} {yCenter} {bw} {bh}")

    return labels
```

```python
def makeImgLabelPair(numTag, img, labelNP, folderName):
    h, w, _ = img.shape
    labelList = labelNP.tolist()

    labels = procRPLTRlabelData(labelList, h, w)

    os.makedirs(f"{folderName}/labels", exist_ok=True)
    os.makedirs(f"{folderName}/images", exist_ok=True)
    
    with open(f"{folderName}/labels/{numTag}.txt", "w") as f:
        f.write("\n".join(labels))
    
    cv2.imwrite(f"{folderName}/images/{numTag}.png", img)
    
```

```python
def runProcessing():
    for i in range(700):  # 0 → 699
        numTag = str(i).zfill(4)
    
        imagePath = fr"C:/PROJECTS/TrafficLightDetectionTutorial/ReplicatorOutput/rgb_{numTag}.png"
        labelPath = fr"C:/PROJECTS/TrafficLightDetectionTutorial/ReplicatorOutput/bounding_box_2d_tight_{numTag}.npy"
        
        img = cv2.imread(imagePath)
        data = np.load(labelPath)

        if data is None or len(data) == 0:
            print(f"No data on {i}")
            continue
    
        makeImgLabelPair(numTag, img, data, "PrepData")

        print(f"---{i}---")
```

```python
runProcessing()
```
