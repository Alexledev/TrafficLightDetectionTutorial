```python
import os
import shutil
import random
from pathlib import Path
```

```python
sourceDir = Path("PrepData")
outputDir = Path("TrainData")

trainRatio = 0.7
valRatio   = 0.2
testRatio  = 0.1

randomSeed = 42
```

```python
random.seed(randomSeed)

imagesDir = sourceDir / "images"
labelsDir = sourceDir / "labels"

imageExts = [".jpg", ".jpeg", ".png"]

images = []
for ext in imageExts:
    images.extend(imagesDir.glob(f"*{ext}"))

images = sorted(images)

if len(images) == 0:
    raise Exception("No images found")

random.shuffle(images)

total    = len(images)
trainEnd = int(total * trainRatio)
valEnd   = trainEnd + int(total * valRatio)

trainFiles = images[:trainEnd]
valFiles   = images[trainEnd:valEnd]
testFiles  = images[valEnd:]

print(f"Total: {total}")
print(f"Train: {len(trainFiles)}")
print(f"Val:   {len(valFiles)}")
print(f"Test:  {len(testFiles)}")
```

```python
def copySplit(files, split):
    imgOut = outputDir / split / "images"
    lblOut = outputDir / split / "labels"

    imgOut.mkdir(parents=True, exist_ok=True)
    lblOut.mkdir(parents=True, exist_ok=True)

    for imgPath in files:
        labelPath = labelsDir / (imgPath.stem + ".txt")

        shutil.copy2(imgPath, imgOut / imgPath.name)

        if labelPath.exists():
            shutil.copy2(labelPath, lblOut / labelPath.name)
        else:
            print(f"Warning: missing label for {imgPath.name}")

copySplit(trainFiles, "train")
copySplit(valFiles,   "val")
copySplit(testFiles,  "test")

print("\n Done. Dataset split at:", outputDir)    
```
